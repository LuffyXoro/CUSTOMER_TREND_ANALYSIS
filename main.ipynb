```python
import pandas as pd

dataframe=pd.read_csv("Data/customer_shopping_behavior.csv")
```

```python
dataframe.head()
dataframe.info()
```

```python
dataframe.describe(include="all")
```

```python
dataframe.isnull().sum()
```

The problem here is, can't directly substitue the overall median ; because it has differnet categories which will create bias in future.


```python
dataframe['Review Rating']=dataframe.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
dataframe.isnull().sum()
```

```python
dataframe.columns=dataframe.columns.str.lower()
```

```python
dataframe.columns=dataframe.columns.str.replace(' ','_')
```

```python
dataframe=dataframe.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
dataframe.columns.tolist()
```

Feature Engg.

```python
dataframe['age_group']=pd.qcut(dataframe['age'],4,labels=['Young Adult', 'Adult', 'Middle-aged', 'Senior'])
```

```python
dataframe[['age_group','purchase_amount']].groupby('age_group').head()
```

```python
dataframe['frequency_of_purchases'].count()
```

```python
frequency_mapping={'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90}
```

```python
dataframe['purchase_frequency_days'] = dataframe['frequency_of_purchases'].map(frequency_mapping)
```

```python
dataframe[['frequency_of_purchases','purchase_frequency_days']].head(10)
```

```python
dataframe[['discount_applied','promo_code_used']].head(10)
```

```python
(dataframe['discount_applied']==dataframe['promo_code_used']).all()  
```

There are no cases where discount is applied without using a promo code, or vice versa. (i.e;Seasonal Discount or BOGO)

```python
# We can conclude that the 'discount_applied' and 'promo_code_used' columns are identical, so we can drop one of them to avoid redundancy.

dataframe=dataframe.drop(columns=['promo_code_used'])
```
